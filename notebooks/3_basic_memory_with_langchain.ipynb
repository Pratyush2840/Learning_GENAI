```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_openai import ChatOpenAI
llm = ChatOpenAI(model="gpt-4o")
```

```python
prompts = [
    {"role":"user", "content":"Hello, My name is pratyush singh."},
    {"role":"ai", "content":'Hello, Pratyush! How can I assist you today?'},
    {"role":"user", "content":"What is My name ?"}
]
```

```python
res = llm.invoke(prompts)
```

```python
res.content
```

##Manage History

```python
history = [] ## [{"role":"user", "content":"Who is pm of india ?"}]
while True:
    query = input("User: ")
    if query.lower() in ["exit", "quit", "bye"]:
        print("Good Bye 👋")
        break

    history.append({"role":"user", "content":query})
    print("User: ", query)


    res = llm.invoke(history)
    history.append({"role":"ai", "content":res.content})
    print("AI: ", res.content, "\n")
    
```
